## **L'importation des Bibliothèque :**


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## **L'importation de data :**


```python
data=pd.read_csv('INFY.csv')
```

## **La divition de data :**


```python
x = data['Prev Close']
y = data['Close']
```

```python
x=x.to_numpy()
y=y.to_numpy()
```

```python
x_train = x[:int(len(data) * 0.8)]
y_train = y[:int(len(data) * 0.8)]

x_test = x[int(len(data) * 0.8):]
y_test = y[int(len(data) * 0.8):]
```

## **La Visualisation De Data :**

```python
plt.scatter(x_train,y_train)
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

# La Development de Modèle de KNN Pour la Regression :

```python
def predict(x_train, y_train, x_test, k):
    pred = []
    for pnt in x_test:
        dist = []
        for i in range(len(x_train)):
            d = abs(x_train[i] - pnt)
            dist.append((d, y_train[i]))

        ptv = sorted(dist)[:k]

        total = 0
        for val in ptv:
            total += val[1]
        pred.append(total / k)

    return pred

y_pred = predict(x_train, y_train, x_test, 3)
```

# L'Evaluation avec test :


## **MSE :**

```python
MSE = np.mean((y_test - y_pred)**2)
MSE
```

## **MAE :**


```python
MAE = np.mean(abs(y_test - y_pred))
MAE
```

## **R2 :**


```python
R2 = 1 - (sum((y_test - y_pred)**2) / sum((y_test - np.mean(y_test))**2))
R2
```
